## **Functions**

```python
# 1. Simple Function
def name() -> None:
    print('Hello, Joseph')

name()
```

```python
# 2. Using Parameters.
def greet(name: str) -> None:
    print(f'Hello, {name}')

# Now you can call the function w/different names
greet('Joseph')
greet('Baker42')
```

```python
# 3. Function parameters default values.
def greet(name: str = 'friend') -> None:
    print(f'Hello, {name}')

greet()             # Output: Hello, friend
greet('Joseph')     # Output: Hello, Joseph
```

### **INPUTS**

```python
# Input Function: name()
def name():
    user_name = input("Please enter your name: ")
    print("Welcome, " + user_name + "!")

name()
```

```python
# Input Function: greet()
def greet():
    name = input("What is your name? ")
    print("Hello, " + name + "!")

greet()
```

```python
# Multiple Inputs: (name & age)
def introduction():
    name = input("What is your name? ")
    age = input("How old are you? ")
    print("Hello " + name + ", you are " + age + " years old!")

introduction()
```

## **Docstrings (""")**

```python
# Single Line Docstrings
def greet(name: str) -> None:
    """Prints a greeting to the specified name."""
    print(f'Hello, {name}')

greet('Joseph')
```

```python
# Multi-Line Docstrings
def greet(name: str) -> None:
    """
    This is a multi-line docstring
    for better explaining the fucntion
    """
    print(f'Hello, {name}')

greet('Joseph')
```

##  **For/While Loop**

```python
# 1. For-Loop: Loop through a list & print each item
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# 2. While-Loop
count = 0
while count < 50:
    print(count)
    count += 2
```

```python
# 3. While-Loop
count = 0
while True:
    print(count)
    count += 1
    if count >= 5:
        break
```

```python
# Prints out only odd numbers - 1,3,5,7,9
for x in range(10):
    # Check if x is even
    if x % 2 == 0:
        continue
    print(x)
```

##  **"Range()" Function**

```python
# Simple: Recursive Function
def name() -> None:
    print('* Hello, Joseph')
for i in range(5):
    name()
```

```python
# Advanced:
def name(user_name: str = 'Joseph') -> None:
    print(f'Hello, {user_name}\n')
for i in range(20):
    print(i)

name()      # will print Hello, Joseph
name('Spenc & Cayden')  # Hello, Spenc & Cayden
```

##  **If/Else Conditional**

```python
# If/Else example
age = 18

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

```python
# elif example
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C")
```
