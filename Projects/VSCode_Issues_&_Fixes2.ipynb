# VS Code: Environments, & Python Debugger Issues:
- Confusion regarding Windows, WSL, & other environment setups in VS Code...

## **Understanding Your VS Code Setup**

1. **Where is Your Code Located?**

   - If your Python project is in a **Windows directory (e.g., `C:\Users\YourName\Projects\MyApp`)**, then running it in WSL **can** cause issues because Windows & WSL handle file paths differently.
   - If your project is inside **WSL (e.g., `/home/yourname/projects/myapp`)**, then VS Code needs to run in WSL mode for proper debugging.

2. **Which Terminal Should You Use?**
   - If you installed **Python on Windows**, then use **Command Prompt** (`cmd`) or **PowerShell**.
   - If you installed **Python in WSL**, then use **Ubuntu (WSL) terminal**.
   - The key is: **Your VS Code terminal must match where your Python is installed!** If you mix them, things break.

---

## **Best Python Setup in VS Code**
There are two **recommended** ways to set up Python in VS Code:

### **🔹 Option 1: Use Windows Python (For Simplicity)**
If you want to use Python installed on Windows:
1. **Install Python for Windows** (if not installed):  
   - Download from [Python.org](https://www.python.org/downloads/).
   - Install & **check box** to **"Add Python to PATH"**.

2. **Set Python Interpreter in VS Code:**
   - Open **VS Code**.
   - Press `Ctrl + Shift + P` & type **"Python: Select Interpreter"**.
   - Choose Python path that looks like:
   
     ```
     C:\Users\YourName\AppData\Local\Programs\Python\Python39\python.exe
     ```

3. **Create & Use a Virtual Environment:**
   
   - Open VS Code terminal (**PowerShell** or **Command Prompt**).
   - Run:

     ```powershell
     python -m venv venv
     ```

   - Activate it:
     - **Command Prompt:** `venv\Scripts\activate`
     - **PowerShell:** 
     `venv\Scripts\Activate.ps1`

   - You’ll now see `(venv)` in terminal.

4. **Run & Debug Python in VS Code:**
   - Open **a Python file** in VS Code.
   - Click **Run and Debug** button.
   - It should now work.

---

### **🔹 Option 2: Use WSL Python (For Linux Development)**
If you want to **develop inside WSL (Ubuntu)**:
1. **Open VS Code in WSL Mode:**
   - Open **VS Code**.
   - Press `Ctrl + Shift + P` & type **"Remote-WSL: New Window"**.
   - This **opens VS Code inside WSL**.

2. **Install Python in WSL:**
   - In **WSL terminal** (inside VS Code), run:

     ```bash
     sudo apt update && sudo apt install python3 python3-venv python3-pip -y
     ```

3. **Create a Virtual Environment in WSL:**
   - Navigate to your project folder inside WSL:

     ```bash
     cd ~/projects/myapp  # Change to your actual project folder
     ```

   - Create a virtual environment:

     ```bash
     python3 -m venv venv
     ```

   - Activate it:

     ```bash
     source venv/bin/activate
     ```

###########################################     

4. **Set Python Interpreter in VS Code (WSL Mode):**
   - Press `Ctrl + Shift + P`, type **"Python: Select Interpreter"**.
   - Choose something like:

     ```
     /home/yourname/projects/myapp/venv/bin/python3
     ```

5. **Run & Debug Python in VS Code (WSL Mode):**
   - Open a **Python file** & click **Run and Debug**.
   - It should now work **inside WSL**.

✅ If you want a **Linux-like development** experience → Use **WSL Python (Option 2).**

## 🚀 Key Differences Between PowerShell & Ubuntu (WSL) in VS Code for Python:

| Feature              | PowerShell (`PS`)                          | Ubuntu Terminal (`WSL`)                    |
|----------------------|---------------------------------|---------------------------------|
| **Command Syntax**    | Uses **Windows PowerShell** commands (`Set-ExecutionPolicy`, `.exe` for executables) | Uses **Linux Bash** commands (`chmod`, `export`, `source`) |
| **Python Installation** | Uses Windows Python (`python.exe`) | Uses Linux Python (`python`) |
| **Virtual Environment Creation** | `python -m venv venv` | `python3 -m venv venv` (often `python3`) |
| **Activate Virtual Environment** | `venv\Scripts\Activate.ps1`  | `source venv/bin/activate` |
| **Deactivate Virtual Environment** | `deactivate` | `deactivate` |
| **Package Management** | `pip install` (uses Windows Python) | `pip install` (uses Linux Python) |
| **File System** | Uses `C:\Users\YourName` paths | Uses `/home/yourname` paths |
| **Path Separator** | Uses `\` (backslash) | Uses `/` (forward slash) |
| **Environment Variables** | `$env:VAR_NAME = "value"` | `export VAR_NAME="value"` |

## **1️⃣ Virtual Environments in PowerShell vs Ubuntu (WSL)**
### **✅ In PowerShell (Windows)**

```powershell
# Create a virtual environment
python -m venv venv

# Activate it
venv\Scripts\Activate.ps1  # PowerShell activation

# Deactivate it (same in both)
deactivate
```

💡 **Note:** You may need to change execution policies if you get an error.

---

### **✅ In Ubuntu (WSL)**

```bash
# Install Python (if not installed)
sudo apt update && sudo apt install python3 python3-venv python3-pip

# Create virtual environment
python3 -m venv venv

# Activate it
source venv/bin/activate  # Linux activation

# Deactivate it
deactivate
```

💡 **Key Difference:**  

- Instead of `venv\Scripts\Activate.ps1`, Linux uses `source venv/bin/activate`.

## **2️⃣ How to Switch Between PowerShell & Ubuntu in VS Code**

In **VS Code**, you can change terminal by:  
1. **Opening terminal** → (`Ctrl + ~` to open it).  
2. **Clicking dropdown arrow** next to terminal name.  
3. **Selecting either**:
   - **PowerShell** → If working w/Windows Python.
   - **WSL (Ubuntu)** → If working w/Linux Python.

To **set Ubuntu (WSL) as default** terminal in VS Code:  
- Open **VS Code Settings** (`Ctrl + ,`)  
- Search for **“terminal.integrated.defaultProfile.windows”**  
- Change value to `"WSL"`.

## **3️⃣ When Should You Use PowerShell vs Ubuntu (WSL)?**
- **Use PowerShell if:**
  - You installed **Python on Windows**.
  - You are working with Windows-based scripts.
  - You need Windows-specific features (`python.exe`, `pip`, `venv\Scripts\Activate.ps1`).

- **Use Ubuntu (WSL) if:**
  - You installed **Python inside WSL**.
  - You're working on a **Linux project** or a **server setup**.
  - You need **Linux commands** (`chmod`, `export`, `source`).
  - You’re using **Python libraries that work better in Linux (like Django, Flask, or ML tools).**

**💡 Best Practice:**  
If your project is Linux-based (e.g., using Ubuntu inside WSL), always use **WSL terminal** in VS Code. Otherwise, if you need Windows compatibility, use PowerShell.

|-------------|----------------|

| `Get-Help <cmd>` | Shows help for a command. |
| `Get-Process` | Lists running processes. |
| `Stop-Process -Name notepad` | Stops Notepad process. |
| `Get-Service` | Lists all services. |
| `Start-Service spooler` | Starts a service (print spooler example). |
| `Set-ExecutionPolicy RemoteSigned` | Allows running scripts. |
